## The profile

```python
from IPython.display import Markdown
from anchor_paradox.profile import load_profile, render_markdown
profile = load_profile(OUT)
Markdown(render_markdown([profile]))
```

## Removal curves
Suppression against capability cost for every disposition and surface.

```python
import matplotlib.pyplot as plt
fig, axes = plt.subplots(1, 3, figsize=(15, 4), sharey=True)
for ax, surface in zip(axes, ["input", "state", "structural"]):
    for name, t in profile["traits"].items():
        pts = t["surfaces"].get(surface, {}).get("points")
        if not pts:
            continue
        ax.plot([p["cost"] for p in pts], [p["suppression"] for p in pts], marker="o", label=name)
    ax.axhline(0.5, ls="--", lw=0.8, color="grey")
    ax.set_title(surface)
    ax.set_xlabel("capability cost")
axes[0].set_ylabel("suppression")
axes[-1].legend(fontsize=8)
plt.tight_layout()
```
